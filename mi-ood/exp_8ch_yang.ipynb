```python
experiment = "MI_hands_8ch"
COMMON_CHANNELS = [
    "FC3", "FCz", "FC4",
    "C3",  "Cz",  "C4",
    "CP3",       "CP4",
]
for dataset in [ "Yang2025_3C" ]:
    print( dataset )
    subjects = b.get_subjects(dataset) 

    for subject in subjects:
        data = None
        for method, pipeline in b.pipelines.items():
            result = {"dataset": dataset, "subject": subject, "experiment": experiment, "method": method}
            if b.result_path(result).exists():
                continue
            if data is None:
                data = b.load_subject_data(dataset, subject, channels=COMMON_CHANNELS)
            train_data, val_data, test_data = data
            result["MI_balanced_accuracy"] = b.MI_balanced_accuracy(train_data, test_data, pipeline)
            b.save_result(result)
            print( f"s{subject} {method} {result['MI_balanced_accuracy']:.3f}" )

    display(b.summarize_results(dataset, experiment).round(3))
```
